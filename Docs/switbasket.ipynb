## Descripción de variables 

**Información del Pedido y Contexto Temporal**

* **`order_id`**
Número únio de identificación de la orden.
* **`year`**
Año de realización de la orden.
* **`month`**
Mes de realización de la orden.
* **`quarter`**
Trimestre en que se realizó la orden.
* **`is_weekend_order`**
True or False -> si la orden se hizo durante un fin de semana.
* **`platform`**
Nombre de la plataforma a través de la que se realizó la orden. 
* **`country`**
País de realización de la orden. 

---

**Perfil y Fidelización del Cliente**

* **`customer_segment`**
Segmento especifico del publico objetivo al que pertenece el cliente. 
* **`customer_age_group`**
Rando de edad al que pertenece el cliente. 
* **`customer_gender`**
Género al que pertenece el cliente. 
* **`acquisition_channel`**
Canal a través del cual se adquirió el cliente. 
* **`device_used`**
Tipo de dispositivo mediante el cual el cliente realizó la orden. 
* **`subscription_member`**
True or False -> el cliente posee membresia?
* **`repeat_purchase_count`**
Conteo de compras repetidas realizadas por el cliente. 
* **`loyalty_points`**
Conteo de puntos de fidelidad del cliente. 

---

**Detalles del Producto y Valores Financieros**

* **`product_category`**
Categoría comercial a la que pertenece el producto comprado.
* **`unit_price_usd`**
Precio por unidad del producto expressado en dólares (USD).
* **`quantity_ordered`**
Número de unidades del producto solicitadas en la orden.
* **`order_value_usd`**
Valor subtotal de la compra antes de aplicar descuentos, impuestos o envío (USD).
* **`discount_pct`**
Porcentaje de descuento aplicado sobre el valor inicial del pedido.
* **`discount_amount_usd`**
Monto total descontado al pedido expressado en dólares (USD).
* **`final_amount_usd`**
Valor de la compra tras aplicar el descuento, antes de impuestos y envío (USD).
* **`shipping_cost_usd`**
Costo cobrado por el servicio de transporte y envío en dólares (USD).
* **`tax_amount_usd`**
Monto calculado por concepto de impuestos aplicados a la compra (USD).
* **`total_paid_usd`**
Monto final total pagado por el cliente incluyendo envío e impuestos (USD).
* **`coupon_used`**
Indica si la compra incluyó el uso de un cupón de descuento (Yes / No).

---

**Logística y Envíos**

* **`payment_method`**
Medio de pago seleccionado por el cliente para concretar la transacción.
* **`shipping_method`**
Tipo o modalidad de servicio de envío seleccionado para la entrega.
* **`delivery_days`**
Número de días que tomó realizar la entrega del pedido.

---

**Navegación Web e Interacción del Usuario**

* **`session_duration_minutes`**
Tiempo total en minutos que duró la sesión de navegación del cliente.
* **`pages_viewed`**
Cantidad de páginas que el cliente visitó dentro de la plataforma durante la sesión.
* **`cart_abandoned_before`**
Indica si el cliente había abandonado un carrito de compras previamente (Yes / No).
* **`wishlist_used`**
Indica si el cliente guardó el producto en la lista de deseos antes de comprar (Yes / No).

---

**Evaluación, Satisfacción y Devoluciones**

* **`product_rating`**
Calificación numérica dada por el cliente al producto adquirido.
* **`review_submitted`**
Indica si el usuario escribió y envió una reseña del producto (Yes / No).
* **`returned`**
Indica si el pedido fue devuelto posteriormente por el cliente (Yes / No).
* **`return_reason`**
Motivo registrado por el cual el cliente realizó la devolución del producto.
* **`customer_satisfaction`**
Nivel cualitativo de satisfacción reportado por el cliente con respecto al servicio o compra.
