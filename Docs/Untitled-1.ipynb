```python
#CLASE SOBRE DATOS ABIERTOS Y APIs: 

#Importar batos en CSV desde una URL:
import pandas as pd
import requests

url = "https://www.datos.gov.co/api/v3/views/mvrj-vu7n/query.json"
response = requests.get(url)
print(response.status_code) # Verificar el código de estado en la respuesta

datos = response.json() # Convertir la respuesta a formato JSON
df = pd.DataFrame(datos) # Crear un Dataframe a partir de los datos
display(df.head(20))  # Mostrar las primeras 20 filas del Dataframe



```
